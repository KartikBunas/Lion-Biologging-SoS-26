```python
file_path = "/content/drive/MyDrive/Lion-Biologging-SoS'26/lion_data.csv"
df = pd.read_csv(file_path)
df.head()
```

```python
#dropping useless columns
try:
  df = df.drop(columns=['timestamp','event-id','visible','eobs:temperature','gps:dop','height-raw','sensor-type','individual-taxon-canonical-name','study-name','study-timezone','tag-local-identifier'])
except:
  df = pd.read_csv(file_path)
  df = df.drop(columns=['timestamp','event-id','visible','eobs:temperature','gps:dop','height-raw','sensor-type','individual-taxon-canonical-name','study-name','study-timezone','tag-local-identifier'])
df.head()
```

```python
#changing names to simplify
df = df.rename(columns={'study-local-timestamp':'local_time','individual-local-identifier':'lion_id','location-lat':"latitude",'location-long':'longitude'})
df.head()
```

```python
#finding and replacing corrupted values
has_nan = df.isna().any().any()
print(has_nan)
#the data we have is already very refined
```

```python
#converting timestamp sting into datetime object
df['local_time'] = pd.to_datetime(df['local_time'])
df.head()
```

```python
#sorting the rows based on priority(id then time)
df = df.sort_values(by=['lion_id','local_time']).reset_index(drop=True)
df.head(15)
```

```python
#saving the clean csv file
df.to_csv("/content/drive/MyDrive/Lion-Biologging-SoS'26/lion_data_clean.csv", index=False)
```
